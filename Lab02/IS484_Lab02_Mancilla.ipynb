```python
import numpy as np

# Definición del mapa como grafo
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}


def bfs(grafo, inicio, objetivo):
  cola = [[inicio]]
  visitados = set()

  while cola:
    camino = cola.pop(0)  # FIFO
    nodo_actual = camino[-1]

    if nodo_actual == objetivo:
      return camino

    if nodo_actual not in visitados:
      visitados.add(nodo_actual)
      for vecino in grafo[nodo_actual]:
        cola.append(camino + [vecino])
  return None


print("Camino BFS:", bfs(mapa, "Almacen", "ZonaNorte"))
```

```python
def dfs(grafo, inicio, objetivo):
  pila = [[inicio]]
  visitados = set()

  while pila:
    camino = pila.pop()  # LIFO
    nodo_actual = camino[-1]

    if nodo_actual == objetivo:
      return camino

    if nodo_actual not in visitados:
      visitados.add(nodo_actual)
      for vecino in grafo[nodo_actual]:
        pila.append(camino + [vecino])
  return None


print("Camino DFS:", dfs(mapa, "Almacen", "ZonaNorte"))
```

```python
obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])

# Producto escalar y distancia euclidiana entre obs1 y obs2
producto = np.dot(obs1, obs2)
dist_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
dist_numpy = np.linalg.norm(obs1 - obs2)

print("Producto escalar:", producto)
print("Distancia manual:", dist_manual)
print("Distancia NumPy:", dist_numpy)
```

```python
nueva_observacion = np.array([20.5, 48])
distancias = []

for i, obs in enumerate(dataset):
  d = np.linalg.norm(nueva_observacion - obs)
  distancias.append(d)
  print(f"Distancia a obs {i}: {d:.4f}")

print("Observación más cercana:", np.argmin(distancias))
```

### Reflexión Final
1. **Medición de Similitud:** La distancia euclidiana mide la separación en un espacio geométrico[cite: 1]; entre menor sea la distancia entre dos vectores, más similares son sus características.
2. **Limitaciones:** Si los datos tienen escalas dispares (ej. ingresos en miles vs. edad en decenas), la variable con magnitudes mayores dominará la distancia[cite: 1]. Para evitar esto, se deben escalar o estandarizar los atributos antes de calcular la distancia.
