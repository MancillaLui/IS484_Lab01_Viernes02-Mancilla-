## Ejercicio 2: Escalamiento de tickets de soporte técnico

### Ficha PEAS

| Componente | Descripción |
| :--- | :--- |
| **Percepción (S)** | Tiempo de espera en minutos (`tiempo_espera_minutos`), nivel de urgencia (`nivel_urgencia`), estado VIP (`cliente_premium`). |
| **Acciones (A)** | Atender en nivel 1 (`"atender en nivel 1"`), Escalar a nivel 2 (`"escalar a nivel 2"`), Escalar a nivel 3 (`"escalar a nivel 3"`). |
| **Entorno (E)** | Sistema de mesa de ayuda (Helpdesk) y cola de atención de tickets de soporte. |
| **Objetivo** | Resolver incidencias eficientemente respetando los acuerdos de nivel de servicio (SLA). |
| **Medida de desempeño (P)** | Tiempo medio de resolución, porcentaje de cumplimiento de SLA y nivel de satisfacción del cliente. |

### Justificación de las reglas
Las reglas priorizan la severidad del problema y los acuerdos de nivel de servicio (SLA). Una urgencia alta requiere solución inmediata por especialistas (Nivel 3). Para urgencias medias o clientes premium, una espera prolongada exige escalar a Nivel 2 para mitigar la insatisfacción, mientras que casos simples de baja urgencia se resuelven en Nivel 1 a menos que el tiempo de espera supere los 45 minutos.

```python
### Código
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    urgencia = nivel_urgencia.lower()
    
    # Criterios para Nivel 3 (Atención crítica / VIP)
    if urgencia == "alta" or (cliente_premium and tiempo_espera_minutos >= 30):
        return "escalar a nivel 3", f"Urgencia alta o cliente premium con espera elevada ({tiempo_espera_minutos} min)"
    
    # Criterios para Nivel 2 (Escalamiento moderado)
    elif urgencia == "media" or cliente_premium or tiempo_espera_minutos >= 45:
        return "escalar a nivel 2", f"Urgencia media, cliente premium o tiempo de espera sostenido ({tiempo_espera_minutos} min)"
    
    # Criterios para Nivel 1 (Atención estándar)
    else:
        return "atender en nivel 1", f"Caso de urgencia baja con tiempo de espera normal ({tiempo_espera_minutos} min)"
```

```python
### Simulación y pruebas
casos_soporte = [
    (10, "alta", False),   # Urgencia alta -> Escalar a Nivel 3
    (35, "baja", True),    # Cliente premium con espera > 30 min -> Escalar a Nivel 3
    (15, "media", False),  # Urgencia media estándar -> Escalar a Nivel 2
    (10, "baja", True),    # Cliente premium urgencia baja -> Escalar a Nivel 2
    (50, "baja", False),   # Espera excesiva regular (> 45 min) -> Escalar a Nivel 2
    (15, "baja", False)    # Caso estándar normal -> Atender en Nivel 1
]

print("--- RESULTADOS DE SIMULACIÓN DE SOPORTE TÉCNICO ---")
for i, (espera, urgencia, premium) in enumerate(casos_soporte, 1):
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(f"Caso {i} [Espera: {espera}m, Urgencia: '{urgencia}', Premium: {premium}]:")
    print(f"  -> Decisión: {accion.upper()} | Motivo: {motivo}\n")
```
